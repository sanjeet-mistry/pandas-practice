```python
import pandas as pd
bios = pd.read_csv('./data/bios.csv')
coffee = pd.read_csv('./warmup-data/coffee.csv')
```

```python
import numpy as np
coffee['Price'] = np.where(coffee['Coffee Type'] == 'Espresso', 6, 10)
```

```python
coffee['Revenue'] = coffee['Price'] * coffee['Units Sold']
```

```python
coffee
```

```python
coffee['yesterday_revenue'] = coffee['Revenue'].shift(2)
```

```python
coffee['revenue_pct_change'] = coffee["Revenue"] / coffee['yesterday_revenue'] * 100
```
