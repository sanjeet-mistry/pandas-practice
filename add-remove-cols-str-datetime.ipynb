```python
import pandas as pd
bios = pd.read_csv("./data/bios.csv")
```

```python
bios.head()
```

```python
bios_copy = bios.copy()
```

```python
bios_copy = bios_copy.drop(columns=["born_region", "NOC"])
```

```python
bios_copy["first_name"] = bios_copy["name"].str.split(" ").str[0]
bios_copy
```

```python
bios_copy["born_date"] = pd.to_datetime(bios_copy["born_date"])
bios_copy.head()
```

```python
bios_copy["year"] = bios_copy["born_date"].dt.year
bios_copy.head()
```

```python
bios_copy.to_csv("./data/bios-new.csv", index=False)
```
