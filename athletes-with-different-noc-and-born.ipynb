```python
import pandas as pd
bios = pd.read_csv("./data/bios.csv")
athletes = pd.read_csv("./data/Olympic_Athlete_Bio.csv")
```

```python
athletes["born"] = athletes["born"].str.strip().pipe(pd.to_datetime, format='%d %B %Y', errors='coerce')
```

```python
bios["born_date"] = pd.to_datetime(bios["born_date"], errors="coerce")
```

```python
bios = bios.dropna(subset=["born_country"])
bios = bios.drop_duplicates()
athletes = athletes.dropna(subset="country_noc")
athletes = athletes.drop_duplicates()
```

```python
merged_df = bios.merge(athletes[["name", "born", "country_noc"]], left_on=["name", "born_date"], right_on=["name", "born"], how="left")
```

```python
merged_df = merged_df[["athlete_id", "name", "born_date", "born_country", "NOC", "country_noc"]]
```

```python
merged_df = merged_df[merged_df["born_country"] != merged_df["country_noc"]]
```

```python
merged_df.to_csv("./output/athletes-with-different-born-and-noc-countries-v2.csv", index=False)
```
