```python
bios = bios.dropna(subset=["height_cm", "weight_kg"])
```

```python
group1 = bios.groupby(['born_country'])['height_cm'].count()
group2 = bios.groupby(['born_country'])['height_cm'].mean()
```

```python
group = pd.DataFrame([group1, group2], index=['count', 'mean'])
```

```python
group = group.transpose()
```
