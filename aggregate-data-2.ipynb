```python
import pandas as pd
coffee = pd.read_csv("./warmup-data/coffee.csv")
```

```python
import numpy as np
coffee['Price'] = np.where(coffee['Coffee Type'] == 'Espresso', 10, 6)
```

```python
coffee.groupby(['Coffee Type'])['Units Sold'].mean()
```

```python
coffee.groupby(['Coffee Type']).agg({'Units Sold': 'sum', 'Price': 'sum'})
```
