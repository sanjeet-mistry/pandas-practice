```python
import pandas as pd
bios = pd.read_csv("./data/bios.csv")
```

```python
bios['born_date'] = pd.to_datetime(bios['born_date'], errors="coerce")
```

```python
bios.groupby(bios['born_date'].dt.year)['athlete_id'].count().sort_values(ascending=False)
```

```python
bios['born_month'] = bios['born_date'].dt.month
bios['born_year'] = bios['born_date'].dt.year
bios.groupby([bios['born_month'], bios['born_year']])['athlete_id'].count().reset_index().sort_values('athlete_id', ascending=False)
```
